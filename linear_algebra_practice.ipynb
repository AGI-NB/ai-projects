```python
import numpy as np

# 第一天：验证 det(AB) = det(A) × det(B)
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print("det(AB):", np.linalg.det(A @ B))
print("det(A)*det(B):", np.linalg.det(A) * np.linalg.det(B))
```

```python
# 第二天：手算一个 3x3 矩阵的特征值，再用 np.linalg.eig 验证
C = np.array([[4, 1, 2], [1, 3, 0], [2, 0, 5]])
eigenvals, eigenvecs = np.linalg.eig(C)
print("特征值:", eigenvals)
```
